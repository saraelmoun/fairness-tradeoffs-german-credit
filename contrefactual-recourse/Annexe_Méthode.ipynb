Annexe A. Spécification formelle détaillée

</h2>


<div style="font-size:1.5em;font-weight:600;margin-bottom:0.4em;">
Annexe A. Construction de PACR-AP, brique par brique
</div>

<p style="line-height:1.95;">
Cette annexe construit la méthode pas à pas, en neuf étapes,
avec définitions formelles et interprétations.
</p>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.1 Étape 1 — Définir ce que signifie « refusé »</b>

<p style="margin-top:0.9em;">
On a besoin d’un critère sans ambiguïté pour dire qu’un individu est refusé,
et donc candidat au recourse, ou favorable.
</p>

<p>On introduit une règle de décision :</p>
</div>

$$
D : [0,1]\times\{0,1\}\to\{0,1\}
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
Elle prend le score et le groupe sensible, puis retourne :
</div>

$$
D(s,a)=
\begin{cases}
0 & \text{refusé} \\
1 & \text{favorable}
\end{cases}
$$

<div style="margin:1.3em 0;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Interprétation.</b>

PACR-AP est agnostique au choix de la règle de décision :
seuil global, seuil par groupe, post-processing de Hardt,
ou règle métier personnalisée.
</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.2 Étape 2 — Restreindre ce qu’on peut modifier</b>

<p style="margin-top:0.9em;">
On ne peut pas demander à un individu de changer son âge ou son sexe.
Il faut donc déclarer explicitement quelles features sont modifiables,
dans quel sens, et entre quelles valeurs.
</p>

<p>On définit un schéma d’actionnabilité :</p>
</div>

$$
S(f)
=
(
\text{mutable}_f,\;
\text{direction}_f,\;
\text{type}_f,\;
\text{valeurs\_autorisées}_f
)
$$

<div style="margin-top:1.3em;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Choix normatif.</b>

Le schéma est externalisé dans un fichier YAML afin de rester négociable.
Trois variantes sont utilisées :
<code>conservateur</code>, <code>modéré</code>, <code>permissif</code>.
</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.3 Étape 3 — Donner un sens à « modifier »</b>

<p style="margin-top:0.9em;">
Un déplacement continu arbitraire dans l’espace des features serait difficile
à interpréter. On préfère donc un petit ensemble d’opérations métier discrètes.
</p>

<p>Un profil est un vecteur de features. L’espace des profils est noté :</p>
</div>

$$
\mathbf{x}\in\mathcal{X}
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
Une action élémentaire est un triplet :
</div>

$$
u=(T_u,\;c_u,\;\pi_u)
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
Ses trois composantes sont :
<ul>
<li>une transformation du profil ;</li>
<li>un coût empirique depuis l’état courant ;</li>
<li>une précondition d’applicabilité.</li>
</ul>
</div>

$$
T_u:\mathcal{X}\to\mathcal{X}
$$

$$
c_u(\mathbf{x})\in\mathbb{R}_+
$$

$$
\pi_u(\mathbf{x})\in\{0,1\}
$$

<div style="margin:1.6em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
<b>Coût d’une action continue</b>

<p style="margin-top:0.9em;">
Pour une feature continue, comme le montant ou la durée,
le coût est standardisé par l’écart-type empirique.
</p>
</div>

$$
c_u(\mathbf{x})=\frac{|\Delta|}{\sigma_f}
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
L’action est interprétée comme un déplacement de plusieurs écarts-types
dans la population observée.
</div>

<div style="margin:1.6em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
<b>Coût d’une action ordinale ou catégorielle</b>

<p style="margin-top:0.9em;">
Pour une feature ordinale ou catégorielle,
le coût dépend de la rareté empirique de la valeur cible.
</p>
</div>

$$
c_u(\mathbf{x})=-\log \widehat{P}(v_{\text{cible}})
$$

$$
\widehat{P}(v_{\text{cible}})
=
\frac{
\#\{i\in\text{train}:x_i^f=v_{\text{cible}}\}
}{
n_{\text{train}}
}
$$

<div style="margin:1.3em 0;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Pourquoi utiliser une pénalité logarithmique ?</b>

C’est la self-information de Shannon :
plus une cible est rare, plus l’action coûte cher.
</div>

$$
-\log(0.1)\approx 2.3
\qquad
\text{contre}
\qquad
-\log(0.5)\approx 0.7
$$

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.4 Étape 4 — Explorer les chemins atteignables</b>

<p style="margin-top:0.9em;">
Combiner plusieurs actions pour basculer la décision demande d’explorer
un espace de séquences. Pour chaque refusé, on construit un graphe d’actions local.
</p>
</div>

$$
G_{\mathbf{x}_0}=(V,E)
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
L’exploration est réalisée par BFS + beam search :
à chaque profondeur, on conserve seulement les états les plus prometteurs,
jusqu’à une profondeur maximale.
</div>

$$
K_{\text{beam}}
\qquad
T_{\max}
$$

<div style="margin-top:1.3em;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Différence avec FACE.</b>

Les nœuds ne sont pas des individus observés,
mais des profils synthétiques générés par application d’actions.
Cette flexibilité impose ensuite de filtrer les états implausibles.
</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.5 Étape 5 — Filtrer les états implausibles</b>

<p style="margin-top:0.9em;">
Un chemin techniquement valide peut passer par des états absurdes.
Chaque état intermédiaire doit donc passer deux filtres conjoints.
</p>
</div>

$$
\text{plausible}(\mathbf{x}_t)
\iff
d_{\text{NN}}(\mathbf{x}_t)\leq\varepsilon
\quad\text{et}\quad
\text{LOF}(\mathbf{x}_t)\geq\rho_{\text{LOF}}
$$

<div style="margin:1.6em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
<b>Filtre marginal</b>

<p style="margin-top:0.9em;">
On mesure la distance au point le plus proche du training.
</p>
</div>

$$
d_{\text{NN}}(\mathbf{x}_t)
=
\min_{\mathbf{z}\in\mathcal{D}_{\text{train}}}
d(\mathbf{x}_t,\mathbf{z})
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
La distance est une distance L1 mixte :
les variables continues sont standardisées,
et les variables catégorielles sont comparées par indicatrice.
</div>

$$
\frac{|x_t^f-z^f|}{\sigma_f}
$$

$$
\mathbb{1}\{x_t^f\neq z^f\}
$$

<div style="margin:1.6em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
<b>Filtre joint : LOF</b>

<p style="margin-top:0.9em;">
Le LOF compare la densité locale autour du point étudié
à celle de ses voisins.
</p>
</div>

$$
\text{LOF}_k(\mathbf{x}_t)
\sim
\frac{
\overline{\text{densité}}
\bigl(\mathbf{y}\in\mathcal{N}_k(\mathbf{x}_t)\bigr)
}{
\text{densité}(\mathbf{x}_t)
}
$$

<div style="margin-top:1.3em;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Convention sklearn.</b>

On utilise <code>negative_outlier_factor</code> :
plus le score est haut, plus le point est typique.

<p style="margin-top:0.9em;">
Les seuils sont calibrés sur le training :
le percentile 95 pour la distance NN,
et le percentile 5 pour le score LOF.
</p>
</div>

$$
\varepsilon = q_{95}(d_{\text{NN}})
$$

$$
\rho_{\text{LOF}} = q_{5}(\text{LOF})
$$

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.6 Étape 6 — Définir un chemin faisable</b>

<p style="margin-top:0.9em;">
Un chemin est une suite de profils.
</p>
</div>

$$
P=(\mathbf{x}_0,\mathbf{x}_1,\ldots,\mathbf{x}_T)
$$

$$
\mathbf{x}_t=T_{u_t}(\mathbf{x}_{t-1})
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
Il est faisable si quatre conditions sont satisfaites :
<ul>
<li>la décision finale est favorable ;</li>
<li>chaque action respecte le schéma d’actionnabilité ;</li>
<li>chaque état intermédiaire passe le double filtre de plausibilité ;</li>
<li>la profondeur reste bornée.</li>
</ul>
</div>

$$
D(f(\mathbf{x}_T),a)=1
$$

$$
T\leq T_{\max}
$$

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.7 Étape 7 — Préférer les chemins robustes au modèle</b>

<p style="margin-top:0.9em;">
Un chemin valide pour le modèle principal peut devenir défavorable
si le modèle est réentraîné sur un échantillon légèrement différent.
On entraîne donc plusieurs modèles bootstrap.
</p>
</div>

$$
f_1,\ldots,f_B
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
Chaque modèle a la même architecture et les mêmes hyperparamètres,
mais il est entraîné sur un rééchantillonnage avec remise du training set.
</div>

$$
\text{RobustValidity}(\mathbf{x}_T)
=
\frac{1}{B}
\sum_{b=1}^{B}
\mathbb{1}\{D(f_b(\mathbf{x}_T),a)=1\}
$$

<div style="margin-top:1.3em;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Limite.</b>

Cette robustesse mesure la sensibilité au train sampling,
mais pas la robustesse à un changement de classe de modèle
comme LR, XGBoost ou Random Forest.
</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.8 Étape 8 — Sélectionner K chemins divers à montrer</b>

<p style="margin-top:0.9em;">
Plusieurs chemins faisables et robustes peuvent exister.
On les résume par six quantités.
</p>
</div>

| Quantité | Définition |
|---|---|
| Coût cumulé | Somme des coûts des actions |
| Longueur | Nombre d’actions |
| Sparsité | Nombre de features distinctes modifiées |
| Plausibilité worst | Pire distance NN rencontrée |
| Marge finale | Dépassement du seuil |
| Robustesse | Fraction des modèles bootstrap favorables |

$$
c(P)=\sum_{t=1}^{T}c_{u_t}(\mathbf{x}_{t-1})
$$

$$
T(P)=T
$$

$$
s(P)=
\left|
\{f:\exists t,\;u_t\text{ modifie }f\}
\right|
$$

$$
d^{\max}_{\text{NN}}(P)
=
\max_{t=0,\ldots,T}
d_{\text{NN}}(\mathbf{x}_t)
$$

$$
m(P)=f(\mathbf{x}_T)-\tau
$$

$$
\rho(P)=\text{RobustValidity}(\mathbf{x}_T)
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
Chaque quantité est normalisée min-max sur les chemins candidats
du même individu.
</div>

$$
\tilde{q}
=
\frac{q-q_{\min}}{q_{\max}-q_{\min}}
\in[0,1]
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
On combine ensuite les termes dans un objectif scalaire.
Les termes à minimiser ont un signe positif,
et les termes à maximiser ont un signe négatif.
</div>

$$
\text{Obj}(P)
=
\lambda_c\tilde{c}
+
\lambda_T\tilde{T}
+
\lambda_s\tilde{s}
+
\lambda_p\tilde{d}^{\max}_{\text{NN}}
-
\lambda_m\tilde{m}
-
\lambda_r\tilde{\rho}
$$

| Coefficient | Valeur | Interprétation |
|---|---:|---|
| lambda coût et lambda plausibilité | 1.0 | priorités principales |
| lambda longueur, marge et robustesse | 0.5 | priorités secondaires |
| lambda sparsité | 0.3 | priorité tertiaire |

<div style="margin-top:1.3em;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Point important.</b>

Le terme de plausibilité porte sur la trajectoire entière,
via le pire état rencontré,
et pas seulement sur le profil final.
</div>

<div style="margin:1.6em 0;padding-left:1.1em;border-left:3px solid #b88a4a;line-height:1.95;">
<b>Contrainte de diversité</b>

<p style="margin-top:0.9em;">
Pour éviter plusieurs variantes triviales du même chemin,
on impose une distance de Jaccard minimale entre chemins sélectionnés.
</p>
</div>

$$
d_J(P_1,P_2)
=
1-
\frac{
|\mathcal{F}_1\cap\mathcal{F}_2|
}{
|\mathcal{F}_1\cup\mathcal{F}_2|
}
$$

$$
d_J\geq 0.5
$$

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
<b>A.9 Étape 9 — Auditer la fairness du recourse par groupe</b>

<p style="margin-top:0.9em;">
Pour juger de l’équité,
on agrège les résultats par groupe sensible.
</p>
</div>

$$
a\in\{0,1\}
$$

$$
\text{coverage}(a),\;
\overline{c}(a),\;
\overline{T}(a),\;
\overline{\rho}(a),\;
\overline{d^{\max}_{\text{NN}}}(a)
$$

<div style="margin:1.4em 0;padding-left:1.1em;border-left:3px solid #7a8fb8;line-height:1.95;">
Les gaps sont calculés dans le sens conventionnel :
</div>

$$
\Delta_M
=
M_{A=0}
-
M_{A=1}
\quad
\text{(jeunes − adultes)}
$$

<div style="margin-top:1.3em;padding:1em 1.15em;background:rgba(255,255,255,0.035);border-left:3px solid #666;border-radius:6px;line-height:1.95;">
<b>Lecture des gaps.</b>

<ul>
<li>Un gap de coût positif signifie que les jeunes paient plus cher leur recours.</li>
<li>Un gap de robustesse négatif signifie que leurs chemins sont moins stables.</li>
<li>Ces métriques complètent DP, EOpp et PP : elles mesurent l’équité du recourse.</li>
</ul>
</div>

$$
\Delta_{\text{cost}}>0
$$

$$
\Delta_{\text{robust}}<0
$$

<div style="margin-top:1.8em;padding-top:0.9em;border-top:1px solid rgba(255,255,255,0.12);font-size:1.02em;line-height:1.95;">
<b>Conclusion.</b>

Avec ces neuf étapes,
PACR-AP est entièrement défini :
on sait qui est refusé,
quelles actions sont autorisées,
comment explorer les chemins,
comment filtrer les trajectoires implausibles,
comment sélectionner les meilleurs recours,
et comment auditer leur équité entre groupes.
</div>
