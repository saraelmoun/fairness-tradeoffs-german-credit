<p style="line-height:1.95;">
Cette annexe résume les principales notions utilisées dans PACR-AP,
leur origine dans la littérature,
et les méthodes dont certaines idées ont été retenues
sans être reprises intégralement.
</p>

<b>Question centrale du recourse contrefactuel</b>

<div style="
text-align:center;
margin:1.2em 0;
font-style:italic;
font-size:1.04em;
">

« Mon dossier de crédit a été refusé par le modèle.
Que dois-je changer pour qu’il soit accepté ? »

</div>

</div>

<p style="line-height:1.95;">
Répondre rigoureusement à cette question nécessite plusieurs notions distinctes.
Pour chacune,
on précise :
</p>

<ul style="line-height:1.9;">
<li>la définition opérationnelle ;</li>
<li>la source littéraire associée ;</li>
<li>l’idée exacte empruntée à cette source.</li>
</ul>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Notion 1 — Contrefactuel basculant</b>

<p style="margin-top:0.9em;">
Un contrefactuel est un profil hypothétique modifié à partir d’un individu refusé,
qui obtiendrait une décision favorable du modèle.
</p>

<p>
Exemple :
</p>

<div style="
text-align:center;
margin:1.2em 0;
font-style:italic;
">

« si vous demandiez 5 000 DM au lieu de 7 000,
le score passerait au-dessus du seuil »

</div>

</div>

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Source.</b>

Wachter, Mittelstadt & Russell (2017)
— <i>Counterfactual Explanations</i>.

<p style="margin-top:0.9em;">
PACR-AP emprunte ici
la définition même de la cible :
trouver une modification minimale
qui bascule la décision.
</p>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Notion 2 — Actionnabilité</b>

<p style="margin-top:0.9em;">
Toutes les modifications ne sont ni possibles ni souhaitables.
</p>

<p>
L’âge,
le sexe,
ou certains éléments historiques
ne doivent pas être modifiés.
</p>

<p>
On introduit donc un schéma déclaratif séparant :
</p>

<ul style="margin-top:0.8em;">
<li>les variables mutables ;</li>
<li>les variables immutables ;</li>
<li>les directions de modification autorisées.</li>
</ul>

</div>

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Source.</b>

Ustun, Spangher & Liu (2019)
— <i>Actionable Recourse</i>.

<p style="margin-top:0.9em;">
PACR-AP emprunte ici
le principe du schéma d’actionnabilité par feature.
</p>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Notion 3 — Diversité multi-contrefactuels</b>

<p style="margin-top:0.9em;">
Un unique contrefactuel optimal est souvent insuffisant.
</p>

<p>
Différents individus peuvent avoir des contraintes personnelles différentes :
certains peuvent réduire le montant demandé,
d’autres préférer allonger la durée,
ou augmenter leur niveau d’épargne.
</p>

<p>
Il est donc préférable de proposer plusieurs alternatives plausibles.
</p>

</div>

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Source.</b>

Mothilal, Sharma & Tan (2020)
— <i>DiCE</i>.

<p style="margin-top:0.9em;">
PACR-AP emprunte ici
l’idée de sélection top-K diversifiée.
</p>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Notion 4 — Chemin d’actions vs saut unique</b>

<p style="margin-top:0.9em;">
Plutôt qu’un saut atomique direct :
</p>

<div style="
text-align:center;
margin:1.2em 0;
font-style:italic;
">

« passez directement de $x_0$ à $x_T$ »

</div>

<p>
PACR-AP décompose le recourse
en une suite explicite d’actions élémentaires :
</p>

<div style="
text-align:center;
margin:1.2em 0;
font-style:italic;
">

« réduisez d’abord le montant de 10 %,
puis améliorez votre épargne d’un cran »

</div>

<p>
Chaque action est interprétable,
et chaque état intermédiaire doit rester réaliste.
</p>

</div>

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Sources.</b>

<ul style="margin-top:0.8em;">

<li>
Karimi, Barthe, Balle & Valera (2020)
— <i>MACE</i> :
énumération discrète d’actions ;
</li>

<li style="margin-top:0.45em;">
Poyiadzi et al. (2020)
— <i>FACE</i> :
plausibilité de trajectoire,
et pas uniquement de l’endpoint.
</li>

</ul>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.6em 0;

line-height:1.95;
">

<b>Notion 5 — Plausibilité hybride</b>

<p style="margin-top:0.9em;">
Un contrefactuel mathématiquement valide
peut rester empiriquement absurde.
</p>

<p>
Par exemple :
un montant élevé combiné à un engagement très faible
peut correspondre à une combinaison quasiment jamais observée.
</p>

<p>
PACR-AP impose donc deux contraintes simultanées :
</p>

<ul style="margin-top:0.8em;">
<li>
le profil doit ressembler à des individus observés
(proximité marginale) ;
</li>

<li style="margin-top:0.45em;">
le profil ne doit pas correspondre à une combinaison rare
de variables
(plausibilité jointe).
</li>
</ul>

</div>

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Sources.</b>

<ul style="margin-top:0.8em;">

<li>
distance marginale héritée du recourse classique ;
</li>

<li style="margin-top:0.45em;">
détection des combinaisons rares via
LOF —
Breunig et al. (2000).
</li>

</ul>

<p style="margin-top:0.9em;">
PACR-AP emprunte ici
le principe des deux filtres conjoints :
proximité locale + densité jointe.
</p>

</div>

<div style="
font-size:1.28em;

">

Méthodes considérées mais non reprises

</div>

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>FACE — chemins entre individus observés uniquement</b>

<p style="margin-top:0.9em;">
FACE impose que les trajectoires passent exclusivement
par des individus réellement observés.
</p>

<p>
PACR-AP assouplit cette contrainte :
les profils synthétiques sont autorisés,
tant qu’ils passent les filtres de plausibilité.
</p>

</div>

<div style="
margin:1.7em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>C-CHVAE et REVISE — modèles génératifs</b>

<p style="margin-top:0.9em;">
Ces approches utilisent des modèles génératifs appris
(autoencodeurs variationnels)
pour produire des contrefactuels plausibles.
</p>

<p>
PACR-AP ne les reprend pas,
car ils sont plus coûteux à entraîner
et moins interprétables.
</p>

<p>
Le filtre LOF capture une partie importante du bénéfice
avec un coût méthodologique beaucoup plus faible.
</p>

</div>

<div style="
margin:1.7em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>DiCE — diversité par DPP</b>

<p style="margin-top:0.9em;">
DiCE propose une diversité fondée sur les
Determinantal Point Processes.
</p>

<p>
PACR-AP préfère une stratégie plus simple :
une sélection gloutonne basée sur la distance de Jaccard
entre ensembles de features modifiées.
</p>

<p>
Cette solution est plus facile à implémenter
et largement suffisante pour :
</p>

</div>

$$
K = 3
$$

<div style="
margin:1.7em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>Causal Recourse — SCM complets</b>

<p style="margin-top:0.9em;">
Karimi, Schölkopf & Valera (2021)
proposent un recourse causal fondé sur un modèle causal structurel complet.
</p>

<p>
Cette approche nécessiterait un SCM crédible pour German Credit,
ce qui dépasse le périmètre du projet.
</p>

<p>
PACR-AP retient néanmoins un principe méthodologique important :
</p>

<div style="
text-align:center;
margin:1.2em 0;
font-style:italic;
">

« cela change la décision du modèle »

</div>

<p>
et jamais :
</p>

<div style="
text-align:center;
margin:1.2em 0;
font-style:italic;
">

« cela vous rendra réellement moins risqué »

</div>

<p>
Autrement dit :
les recommandations restent prédictives,
et non causales.
</p>

</div>

<div style="
margin-top:2.2em;
padding-top:1em;
border-top:1px solid rgba(255,255,255,0.12);
font-size:1.02em;
line-height:1.95;
">

<b>Synthèse.</b>

PACR-AP n’est donc pas une méthode entièrement nouvelle construite ex nihilo,
mais une combinaison cohérente de plusieurs idées issues de la littérature :

<ul style="margin-top:0.9em;">
<li>contrefactuels basculants ;</li>
<li style="margin-top:0.45em;">actionnabilité déclarative ;</li>
<li style="margin-top:0.45em;">chemins d’actions plausibles ;</li>
<li style="margin-top:0.45em;">diversité multi-recours ;</li>
<li style="margin-top:0.45em;">plausibilité hybride locale et jointe.</li>
</ul>

<p style="margin-top:0.9em;">
La contribution principale réside dans leur intégration
dans un pipeline unique orienté audit fairness of recourse.
</p>

</div>
