<h2 style="
color:#b07a4f;
font-weight:650;
letter-spacing:0.01em;
">
Annexe C. Mining des magnitudes depuis le training set

</h2>


<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
font-size:1.45em;
font-weight:600;
margin-bottom:0.35em;
">


</div>

<p style="line-height:1.95;">
Cette annexe détaille comment les magnitudes d’actions sont
apprises depuis les données,
plutôt que choisies arbitrairement à la main.
</p>

<p style="line-height:1.95;">
L’objectif est d’éviter des actions artificielles du type :
</p>

<div style="
text-align:center;
margin:1.2em 0;
font-style:italic;
">

« réduire le montant de 10 % »

</div>

<p style="line-height:1.95;">
sans justification empirique.
Les actions proposées doivent au contraire refléter
ce qu’on observe réellement chez les emprunteurs
ayant basculé de refusé à favorable dans les données.
</p>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Idée générale</b>

<p style="margin-top:0.9em;">
Si l’on veut savoir de combien il faut typiquement réduire un montant
pour changer la décision du modèle,
le plus naturel est de comparer :
</p>

<ul style="margin-top:0.8em;">
<li>des individus refusés ;</li>
<li style="margin-top:0.45em;">
leurs voisins favorables les plus proches.
</li>
</ul>

<p style="margin-top:0.9em;">
Les écarts réellement observés deviennent ensuite
les magnitudes candidates des actions.
</p>

</div>

<div style="
margin-top:1.4em;
padding:1em 1.15em;
background:rgba(255,255,255,0.035);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Intuition.</b>

Le mining ne cherche pas :
</p>

<div style="
text-align:center;
margin:1.1em 0;
font-style:italic;
">

« quelles actions causent réellement un meilleur remboursement »

</div>

<p>
mais :
</p>

<div style="
text-align:center;
margin:1.1em 0;
font-style:italic;
">

« quelles différences caractérisent empiriquement
les profils favorables voisins »

</div>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
font-size:1.25em;
font-weight:600;
margin:2em 0 1.2em 0;
">

Pipeline détaillé

</div>

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Étape 1 — Séparer refusés et favorables</b>

<p style="margin-top:0.9em;">
Le training set est découpé en deux sous-ensembles :
</p>

</div>

$$
\mathcal{R}
=
\{x_i : y_i = 0\}
$$

$$
\mathcal{F}
=
\{x_j : y_j = 1\}
$$

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<ul style="margin-top:0.2em;">
<li>
$\mathcal{R}$ contient les profils refusés ;
</li>

<li style="margin-top:0.45em;">
$\mathcal{F}$ contient les profils favorables.
</li>
</ul>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Étape 2 — Définir une distance mixte</b>

<p style="margin-top:0.9em;">
Chaque refusé est comparé à chaque favorable
à l’aide d’une distance L1 mixte.
</p>

<p>
Pour les variables continues,
les écarts sont standardisés par l’écart-type empirique :
</p>

</div>

$$
\frac{|x_i^f - x_j^f|}{\sigma_f}
$$

<div style="
margin:1.4em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

Pour les variables catégorielles,
on ajoute simplement :
</div>

$$
\mathbb{1}\{x_i^f \neq x_j^f\}
$$

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Interprétation.</b>

Deux individus sont proches s’ils diffèrent peu
sur les variables numériques
et partagent la plupart des catégories importantes.

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Étape 3 — Chercher les voisins favorables</b>

<p style="margin-top:0.9em;">
Pour chaque refusé,
on récupère les :
</p>

</div>

$$
k = 10
$$

<div style="
margin:1.4em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

voisins favorables les plus proches.
</div>

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

On obtient donc un ensemble de paires :

<ul style="margin-top:0.8em;">
<li>(refusé, voisin favorable)</li>
</ul>

à partir desquelles seront extraites
les magnitudes observées.

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Étape 4 — Collecter les écarts feature par feature</b>

<p style="margin-top:0.9em;">
Les différences sont ensuite enregistrées séparément
pour chaque variable mutable,
en respectant les contraintes du schéma d’actionnabilité :
</p>

<ul style="margin-top:0.8em;">
<li><code>increase_only</code></li>
<li><code>decrease_only</code></li>
</ul>

</div>

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>Variables continues</b>

<p style="margin-top:0.9em;">
On stocke le pourcentage de variation observé :
</p>

</div>

$$
\text{pct}
=
\frac{|v_f - v_r|}{|v_r|}
$$

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>Variables ordinales</b>

<p style="margin-top:0.9em;">
On stocke le nombre de crans séparant les deux états :
</p>

</div>

$$
\text{step}
=
\text{idx}(v_f)-\text{idx}(v_r)
$$

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>Variables catégorielles</b>

<p style="margin-top:0.9em;">
On stocke directement la valeur cible observée.
</p>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Étape 5 — Filtre de support statistique</b>

<p style="margin-top:0.9em;">
Une feature n’est retenue
que si elle a accumulé suffisamment d’observations
dans les paires précédentes.
</p>

</div>

$$
\text{min\_support}=30
$$

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

<b>Pourquoi ce filtre ?</b>

Il permet d’éliminer
les coïncidences statistiques fragiles
et les transitions observées trop rarement.

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
margin:1.5em 0;
padding-left:1.1em;
border-left:3px solid #7a8fb8;
line-height:1.95;
">

<b>Étape 6 — Discrétiser en actions canoniques</b>

<p style="margin-top:0.9em;">
Les écarts bruts observés sont ensuite convertis
en un petit ensemble d’actions standardisées.
</p>

</div>

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>Variables continues</b>

<p style="margin-top:0.9em;">
On retient les percentiles :
</p>

</div>

$$
q_{25},
\quad
q_{50},
\quad
q_{75}
$$

<div style="
margin:1.4em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

des pourcentages observés,
avec bornage dans l’intervalle :
</div>

$$
[2\%,\,50\%]
$$

<div style="
margin-top:1.3em;
padding:1em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #666;
border-radius:6px;
line-height:1.95;
">

Le bornage évite :
<ul style="margin-top:0.8em;">
<li>les actions triviales ;</li>
<li style="margin-top:0.45em;">les transformations irréalistes.</li>
</ul>

</div>

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>Variables ordinales</b>

<p style="margin-top:0.9em;">
On conserve les deux valeurs de step
les plus fréquentes,
avec :
</p>

</div>

$$
|\text{step}|\leq 2
$$

<div style="
margin:1.6em 0;
padding-left:1.1em;
border-left:3px solid #b88a4a;
line-height:1.95;
">

<b>Variables catégorielles</b>

<p style="margin-top:0.9em;">
On retient les trois valeurs cibles
les plus fréquentes.
</p>

</div>

<hr style="margin:3em 0;border:none;border-top:1px solid rgba(255,255,255,0.10);">

<div style="
font-size:1.25em;
font-weight:600;
margin:2em 0 1.2em 0;
">

Ce que produit le mining sur German Credit

</div>

<p style="line-height:1.95;">
La cellule suivante imprime :
</p>

$$
\texttt{MINING\_REPORT}
$$

<p style="line-height:1.95;">
qui contient,
pour chaque feature mutable du schéma conservateur :
</p>

<ul style="line-height:1.9;">
<li>le nombre de transitions observées ;</li>
<li style="margin-top:0.45em;">
les statistiques descriptives des écarts observés ;
</li>
<li style="margin-top:0.45em;">
les magnitudes finalement retenues comme actions canoniques.
</li>
</ul>

<div style="
margin-top:1.6em;
padding:0.95em 1.15em;
background:rgba(255,255,255,0.03);
border-left:3px solid #8d6cab;
border-radius:6px;
line-height:1.95;
font-size:0.98em;
">

<b>Point important.</b>

Les transitions observées restent des corrélations,
et non des interventions causales.

<p style="margin-top:0.9em;">
Le mining permet d’ancrer les magnitudes
dans les données réelles,
mais ne résout pas la non-causalité fondamentale
du recourse prédictif.
</p>

</div>
